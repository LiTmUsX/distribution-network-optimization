```python
active_mask = warehouse_master["is_active"] == 1

active_cost_min = warehouse_master.loc[
    active_mask,
    "Cost/unit"
].min()

active_cost_max = warehouse_master.loc[
    active_mask,
    "Cost/unit"
].max()

active_capacity_min = warehouse_master.loc[
    active_mask,
    "Daily Capacity"
].min()

active_capacity_max = warehouse_master.loc[
    active_mask,
    "Daily Capacity"
].max()

print("=" * 60)
print("OBSERVED ACTIVE-FLEET BENCHMARK")
print("=" * 60)

print()
print("Active plants :", active_mask.sum())

print()
print(
    f"Active Cost/unit range     : "
    f"{active_cost_min:.6f} - {active_cost_max:.6f}"
)

print(
    f"Active Daily Capacity range: "
    f"{active_capacity_min} - {active_capacity_max}"
)

print()
print("Observed active plants:")

print(
    warehouse_master.loc[
        active_mask,
        [
            "Plant Code",
            "Cost/unit",
            "Daily Capacity"
        ]
    ]
    .sort_values("Plant Code")
    .to_string(index=False)
)
```

```python
def screen_candidate(row):
    if row["is_active"] == 1:
        return "active"

    fails_cost = row["Cost/unit"] > active_cost_max
    fails_capacity = row["Daily Capacity"] < active_capacity_min

    if fails_cost or fails_capacity:
        return "excluded"

    return "candidate"


warehouse_master["screening_status"] = (
    warehouse_master.apply(
        screen_candidate,
        axis=1
    )
)

print("=" * 60)
print("PLANT SCREENING RESULTS")
print("=" * 60)

print()
print("Screening counts:")
print(
    warehouse_master["screening_status"]
    .value_counts()
)

print()
print("All plants:")

print(
    warehouse_master[
        [
            "Plant Code",
            "Cost/unit",
            "Daily Capacity",
            "is_active",
            "screening_status"
        ]
    ]
    .sort_values("Plant Code")
    .to_string(index=False)
)

print()
print("Excluded plants:")

print(
    warehouse_master.loc[
        warehouse_master["screening_status"] == "excluded",
        [
            "Plant Code",
            "Cost/unit",
            "Daily Capacity"
        ]
    ]
    .sort_values("Plant Code")
    .to_string(index=False)
)

print()
print("Candidate plants:")

print(
    warehouse_master.loc[
        warehouse_master["screening_status"] != "excluded",
        [
            "Plant Code",
            "Cost/unit",
            "Daily Capacity",
            "is_active",
            "screening_status"
        ]
    ]
    .sort_values("Plant Code")
    .to_string(index=False)
)
```

```python
PLANNING_HORIZON_DAYS = 5

final_candidate_pool = warehouse_master[
    warehouse_master["screening_status"] != "excluded"
].copy()
final_candidate_pool["Capacity_wave"] = (
    final_candidate_pool["Daily Capacity"]
    * PLANNING_HORIZON_DAYS
)

total_capacity_wave = final_candidate_pool["Capacity_wave"].sum()
total_demand_wave = customer_demand["order_count"].sum()

print("=" * 60)
print("CAPACITY CONVERSION")
print("=" * 60)

print(
    f"Total wave capacity : {total_capacity_wave:,}"
)

print(
    f"Total wave demand   : {total_demand_wave:,}"
)

print(
    f"Capacity / Demand   : "
    f"{total_capacity_wave / total_demand_wave:.2f}x"
)


tier1_cities = {
    "Mumbai",
    "Delhi NCR",
    "Bengaluru",
    "Chennai",
    "Kolkata",
    "Hyderabad",
    "Pune"
}

RENT_PER_SQFT_MONTH = {
    "tier1": 35,
    "tier2": 18
}

FOOTPRINT_SQFT_PER_ORDER_CAPACITY = 50

USD_TO_INR = 95

final_candidate_pool["city_tier"] = (
    final_candidate_pool["city"]
    .apply(
        lambda c: "tier1"
        if c in tier1_cities
        else "tier2"
    )
)

final_candidate_pool["footprint_sqft"] = (
    final_candidate_pool["Daily Capacity"]
    * FOOTPRINT_SQFT_PER_ORDER_CAPACITY
)

final_candidate_pool["rent_per_sqft_month"] = (
    final_candidate_pool["city_tier"]
    .map(RENT_PER_SQFT_MONTH)
)

final_candidate_pool["fixed_cost_annual"] = (
    final_candidate_pool["footprint_sqft"]
    * final_candidate_pool["rent_per_sqft_month"]
    * 12
)
final_candidate_pool["Cost_per_unit_INR"] = (
    final_candidate_pool["Cost/unit"]
    * USD_TO_INR
)

print()
print("=" * 60)
print("FIXED COST + VARIABLE COST SETUP")
print("=" * 60)

print()
print(
    "Tier distribution:"
)

print(
    final_candidate_pool["city_tier"]
    .value_counts()
    .to_dict()
)

print()
print(
    final_candidate_pool[
        [
            "Plant Code",
            "is_active",
            "city",
            "city_tier",
            "Daily Capacity",
            "Capacity_wave",
            "footprint_sqft",
            "rent_per_sqft_month",
            "fixed_cost_annual",
            "Cost_per_unit_INR"
        ]
    ]
    .sort_values("Cost_per_unit_INR")
    .to_string(index=False)
)
```

```python
order_weights = order_list[["Plant Code", "Weight"]].drop_duplicates().copy()

order_weights["has_2day_rate"] = order_weights.apply(
    lambda r: (
        (
            (two_day_options["Plant Code"] == r["Plant Code"]) &
            (two_day_options["minm_wgh_qty"] <= r["Weight"]) &
            (r["Weight"] <= two_day_options["max_wgh_qty"])
        ).any()
    ),
    axis=1
)

print("=" * 60)
print("2-DAY WEIGHT-BAND COVERAGE")
print("=" * 60)

print()
print("Unique plant-weight combinations :", len(order_weights))
print(
    "Covered by 2-day tariff           :",
    order_weights["has_2day_rate"].sum()
)
print(
    "Not covered                       :",
    (~order_weights["has_2day_rate"]).sum()
)

print()

coverage_by_plant = (
    order_weights
    .groupby("Plant Code")["has_2day_rate"]
    .agg(
        unique_weights="count",
        covered_weights="sum"
    )
)

coverage_by_plant["coverage_pct"] = (
    coverage_by_plant["covered_weights"]
    / coverage_by_plant["unique_weights"]
    * 100
)

print("Coverage by plant:")
print(coverage_by_plant.to_string())

print()

uncovered = order_weights[
    ~order_weights["has_2day_rate"]
]

print("Uncovered plant-weight combinations:")
print(uncovered.head(20).to_string(index=False))
```

```python
def get_freight_cost(plant_code, weight):
    options = two_day_options[
        two_day_options["Plant Code"] == plant_code
    ]

    valid = options[
        (options["minm_wgh_qty"] <= weight) &
        (weight <= options["max_wgh_qty"])
    ].copy()

    valid["shipment_cost"] = np.maximum(
        valid["minimum cost"],
        valid["rate"] * weight
    )

    return valid["shipment_cost"].min()

sample_orders = order_list.sample(
    10,
    random_state=42
).copy()

sample_orders["freight_cost_usd"] = sample_orders.apply(
    lambda r: get_freight_cost(
        r["Plant Code"],
        r["Weight"]
    ),
    axis=1
)

print("=" * 60)
print("FREIGHT COST VALIDATION")
print("=" * 60)

print()
print(
    sample_orders[
        [
            "Order ID",
            "Plant Code",
            "Weight",
            "freight_cost_usd"
        ]
    ].to_string(index=False)
)

print()
print(
    "Missing freight costs :",
    sample_orders["freight_cost_usd"].isna().sum()
)

print(
    "Negative freight costs :",
    (sample_orders["freight_cost_usd"] < 0).sum()
)
```

```python
from math import radians, sin, cos, asin, sqrt

def haversine_km(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(
        radians,
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        sin(dlat / 2) ** 2
        + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    )

    return 6371.0 * 2 * asin(sqrt(a))

distance_rows = []

for port, (port_lat, port_lon) in port_coordinates.items():
    for _, zone in customer_demand_regional.iterrows():
        distance_rows.append({
            "Port": port,
            "city": zone["city"],
            "distance_km": haversine_km(
                port_lat,
                port_lon,
                zone["latitude"],
                zone["longitude"]
            )
        })

port_zone_distance = pd.DataFrame(distance_rows)

print("=" * 60)
print("PORT → REGIONAL ZONE DISTANCE MATRIX")
print("=" * 60)

print()
print("Distance rows :", len(port_zone_distance))
print("Unique ports  :", port_zone_distance["Port"].nunique())
print("Unique zones  :", port_zone_distance["city"].nunique())

print()
print("Missing distances :",
      port_zone_distance["distance_km"].isna().sum())

print()
print("Distance range :",
      f"{port_zone_distance['distance_km'].min():.2f}",
      "-",
      f"{port_zone_distance['distance_km'].max():.2f}",
      "km")

print()
print("Sample distances:")
print(
    port_zone_distance
    .head(20)
    .to_string(index=False)
)
```

```python
CIRCUITY_FACTOR = 1.3

port_zone_distance["modeled_distance_km"] = (
    port_zone_distance["distance_km"] * CIRCUITY_FACTOR
)

print("=" * 60)
print("PORT → REGIONAL ZONE MODELED DISTANCE")
print("=" * 60)

print()
print("Circuity factor :", CIRCUITY_FACTOR)

print()
print(
    "Straight-line distance range :",
    f"{port_zone_distance['distance_km'].min():.2f}",
    "-",
    f"{port_zone_distance['distance_km'].max():.2f}",
    "km"
)

print(
    "Modeled distance range       :",
    f"{port_zone_distance['modeled_distance_km'].min():.2f}",
    "-",
    f"{port_zone_distance['modeled_distance_km'].max():.2f}",
    "km"
)

print()
print("Sample:")
print(
    port_zone_distance[
        [
            "Port",
            "city",
            "distance_km",
            "modeled_distance_km"
        ]
    ]
    .head(20)
    .to_string(index=False)
)

print()
print(
    "Missing modeled distances :",
    port_zone_distance["modeled_distance_km"].isna().sum()
)
```

```python
PORT_ZONE_RATE_INR_PER_TONNE_KM = 5

customer_zone_map = customer_demand[
    ["Customer", "city"]
].copy()

order_list = order_list.merge(
    customer_zone_map,
    on="Customer",
    how="left",
    validate="many_to_one"
)

order_list = order_list.merge(
    port_zone_distance[
        [
            "Port",
            "city",
            "modeled_distance_km"
        ]
    ],
    left_on=["Origin Port", "city"],
    right_on=["Port", "city"],
    how="left",
    validate="many_to_one"
)

order_list["weight_tonnes"] = (
    order_list["Weight"] / 1000
)

order_list["port_zone_cost_inr"] = (
    order_list["weight_tonnes"]
    * order_list["modeled_distance_km"]
    * PORT_ZONE_RATE_INR_PER_TONNE_KM
)

print("=" * 60)
print("PORT → ZONE FREIGHT COST")
print("=" * 60)

print()
print("Rate :", f"₹{PORT_ZONE_RATE_INR_PER_TONNE_KM}/tonne-km")

print()
print("Total orders :", len(order_list))

print(
    "Orders with zone mapping :",
    order_list["city"].notna().sum()
)

print(
    "Orders with distance :",
    order_list["modeled_distance_km"].notna().sum()
)

print(
    "Missing port-zone costs :",
    order_list["port_zone_cost_inr"].isna().sum()
)

print()
print(
    "Total port → zone cost :",
    f"₹{order_list['port_zone_cost_inr'].sum():,.2f}"
)

print(
    "Average port → zone cost/order :",
    f"₹{order_list['port_zone_cost_inr'].mean():,.2f}"
)

print()
print("Sample:")
print(
    order_list[
        [
            "Order ID",
            "Customer",
            "Origin Port",
            "city",
            "Weight",
            "modeled_distance_km",
            "port_zone_cost_inr"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```
